# R_GLAB 386.3.9 - Adding Columns to DataFrames

```python
import pandas as pd
import numpy as np
```

```python
data = {
    'Name': ['Jane', 'Princi', 'James', 'Fadi'],
    'Height': [5.1, 6.2, 5.1, 5.2],
    'Qualifications': ['Msc', 'MA', 'Msc', 'Msc'],
    'Score 1': [50, 86, 77, 45],
    'Score 2': [50, 96, 60, 30]
}

df = pd.DataFrame(data)
```

```python
print(df)
```

```python
address = ['NYC', 'NJ', 'PA','CA']

df['address'] = address

print(df)
```

```python
# create a new total score column

df['Total Score'] = df['Score 1'] + df['Score 2']

print(df)
```

```python
# create average score column

df['Average Score'] = df['Score 1'] + df['Score 2']/ 2
```
